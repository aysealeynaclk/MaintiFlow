AI4I 2020 Predictive Maintenance Dataset üzerinde ilk inceleme.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option('display.max_columns', None)
df = pd.read_csv('../data/ai4i2020.csv')
df.head()
```

```python
df.isnull().sum()
```

## Sınıf dengesizliği: Machine failure oranı

```python
counts = df['Machine failure'].value_counts()
pct = df['Machine failure'].value_counts(normalize=True) * 100
print('Adet:\n', counts)
print('\nYuzde:\n', pct.round(2))
```

```python
fig, ax = plt.subplots(figsize=(4,4))
sns.countplot(x='Machine failure', data=df, ax=ax)
ax.set_title('Machine Failure Dagilimi (0=Saglam, 1=Ariza)')
for p in ax.patches:
    ax.annotate(f'{int(p.get_height())}', (p.get_x()+p.get_width()/2, p.get_height()), ha='center', va='bottom')
plt.tight_layout()
plt.savefig('../models/eda_class_balance.png', dpi=100)
plt.show()
```

## Arıza tipi kırılımı (TWF/HDF/PWF/OSF/RNF)
Not: Bir satırda birden fazla arıza tipi bayrağı aynı anda 1 olabilir mi, kontrol ediyoruz.

```python
failure_cols = ['TWF', 'HDF', 'PWF', 'OSF', 'RNF']
print('Her tipin toplam görülme sayısı:')
print(df[failure_cols].sum())
print('\nBir satırda aynı anda kaç tip arıza bayrağı 1 olabiliyor (dağılım):')
print(df[failure_cols].sum(axis=1).value_counts().sort_index())
```

## Sensör değerlerinin dağılımı

```python
sensor_cols = ['Air temperature [K]', 'Process temperature [K]', 'Rotational speed [rpm]', 'Torque [Nm]', 'Tool wear [min]']
fig, axes = plt.subplots(2, 3, figsize=(15, 8))
for ax, col in zip(axes.flat, sensor_cols):
    sns.histplot(df[col], kde=True, ax=ax)
    ax.set_title(col)
axes.flat[-1].axis('off')
plt.tight_layout()
plt.savefig('../models/eda_sensor_distributions.png', dpi=100)
plt.show()
```

## Type (ürün kalitesi: L/M/H) dağılımı ve arıza ile ilişkisi

```python
print(df['Type'].value_counts())
print('\nType bazinda ariza orani (%):')
print((df.groupby('Type')['Machine failure'].mean() * 100).round(2))
```
